```python
import yaml
from dotenv import load_dotenv
from smolagents import ToolCallingAgent, OpenAIServerModel, tool, DuckDuckGoSearchTool, HfApiModel
import requests
import os
import warnings

warnings.filterwarnings(action="ignore")
```

```python
load_dotenv()
```

```python
with open("/mnt/d/work/smolagents/src/smolagents/prompts/tool_calling_agent2.yaml", "r") as file:
    prompt_templates = yaml.safe_load(file)
```

```python
print(prompt_templates)
```

```python
@tool
def get_weather(location: str) -> dict:
    """this function takes the location and returns the weather statistics:
    Args:
        location: place for which we want the weather statistics like temperature, weather description,wind speed, wind degree, wind direction,pressure
                  precipitation,humidity,cloud cover ,feels like ,uv index and visibility.
    """
    access_key = os.getenv("WEATHER_API_KEY")
    endpoint = f"http://api.weatherstack.com/current?access_key={access_key}&query={location}"
    resp = requests.get(endpoint)
    return resp.json()
```

```python
model = OpenAIServerModel(model_id="gpt-4")
```

```python
agent = ToolCallingAgent(model=model, tools=[get_weather, DuckDuckGoSearchTool()], prompt_templates=prompt_templates)
```

```python
agent.run("what is the current temperature in Agra India?")
```

```python
agent.run(task="who was the second prime minister of india?")
```
